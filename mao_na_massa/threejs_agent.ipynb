```python
from pydantic_ai import Agent
import nest_asyncio
nest_asyncio.apply()
```

```python
from google.colab import userdata
GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')
import os
os.environ['GEMINI_API_KEY'] = GEMINI_API_KEY
```

```python
from pydantic_ai.models.gemini import GeminiModel
modelo = GeminiModel('gemini-2.0-flash', provider='google-gla')
```

```python
agente = Agent(
          model=modelo,
          system_prompt=f'''
          Você é um gerador de páginas web especialista em three.js.
          Ao ser requisitado você deve criar o conteúdo de uma página html em um único arquivo
          fazendo uso de three.js para uma visualização interessante do tema
          proposto. Retorne apenas o código sem qualquer texto ou comentário adicional.

      '''
      )
```

```python
resultado = agente.run_sync('Gere uma página para a apresentação da Gatito Petshop em tons escuros e elementos 3D.')
```
